# 📊 Income Distribution & Household Demographics Analysis

This project analyzes a synthetic dataset to explore descriptive statistics, distributions, and insights.

```
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

```
# Generate Dataset (150+ records)
n = 200

data = pd.DataFrame({
    "Household_ID": [f"H{i}" for i in range(1, n+1)],
    "Age_of_Household_Head": np.random.randint(25, 70, n),
    "Household_Income": np.random.normal(50000, 15000, n).astype(int),
    "Education_Level": np.random.choice(["Primary", "Secondary", "Graduate", "Post-Graduate"], n),
    "Family_Size": np.random.randint(2, 8, n),
    "Owns_House": np.random.choice(["Yes", "No"], n),
    "Urban_Rural": np.random.choice(["Urban", "Rural"], n)
})

data.head()
```

```
# Identify Numerical and Categorical Columns
numerical_cols = data.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = data.select_dtypes(include=['object']).columns

print("Numerical Columns:", numerical_cols)
print("Categorical Columns:", categorical_cols)
```

```
# Mean, Median, Mode
print("Income Mean:", data['Household_Income'].mean())
print("Income Median:", data['Household_Income'].median())
print("Income Mode:", data['Household_Income'].mode()[0])

print("\nAge Mean:", data['Age_of_Household_Head'].mean())
print("Age Median:", data['Age_of_Household_Head'].median())
print("Age Mode:", data['Age_of_Household_Head'].mode()[0])
```

```
# Range, Variance, Standard Deviation
income = data['Household_Income']

print("Range:", income.max() - income.min())
print("Variance:", income.var())
print("Standard Deviation:", income.std())
```

```
# IQR
Q1 = income.quantile(0.25)
Q3 = income.quantile(0.75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```
# Histogram + KDE
plt.figure()
sns.histplot(data['Household_Income'], kde=True)
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()
```

```
# Boxplot: Income vs Education
plt.figure()
sns.boxplot(x='Education_Level', y='Household_Income', data=data)
plt.title("Income by Education Level")
plt.show()
```

```
# Skewness & Kurtosis
print("Skewness:", income.skew())
print("Kurtosis:", income.kurt())
```
